```python
!pip install -q xgboost seaborn

import pandas as pd, numpy as np
import matplotlib.pyplot as plt, seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestRegressor
from xgboost import XGBRegressor
from sklearn.metrics import classification_report, confusion_matrix
from sklearn.metrics import mean_squared_error, r2_score

# 1. Load the dataset
url = "https://raw.githubusercontent.com/JackRoss10089/STA-321/main/Admission_Predict.csv"
df = pd.read_csv(url)
df.rename(columns={'Chance of Admit ': 'Chance_of_Admit'}, inplace=True)
df.drop('Serial No.', axis=1, inplace=True)

print("Dataset preview:")
print(df.head())

# 2. Preview distribution
plt.figure(figsize=(5,3))
sns.histplot(df['Chance_of_Admit'], kde=True)
plt.title("Chance_of_Admit Distribution")
plt.show()

# 3. Binary classification target
df['Admit'] = (df['Chance_of_Admit'] >= 0.8).astype(int)
print("Admit counts:\n", df['Admit'].value_counts())

# 4. Features and targets
X = df.drop(['Chance_of_Admit','Admit'], axis=1)
y_class = df['Admit']
y_reg = df['Chance_of_Admit']

# 5. Train/test split
X_train, X_test, yc_train, yc_test, yr_train, yr_test = train_test_split(
    X, y_class, y_reg, test_size=0.2, random_state=42)

# 6. Define models
models_clf = {
    "Logistic Regression": LogisticRegression(max_iter=1000)
}
models_reg = {
    "Random Forest": RandomForestRegressor(n_estimators=100, random_state=42),
    "XGBoost": XGBRegressor(n_estimators=100, learning_rate=0.1, random_state=42)
}

# 7. Train & evaluate classification model
clf = list(models_clf.values())[0]
clf.fit(X_train, yc_train)
yp_clf = clf.predict(X_test)
print("\n📈 Logistic Regression Classification Report:\n", classification_report(yc_test, yp_clf))
cm = confusion_matrix(yc_test, yp_clf)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=['No Admit','Admit'], yticklabels=['No Admit','Admit'])
plt.title("Confusion Matrix - Logistic Regression")
plt.show()

# 8. Train & evaluate regression models
for name, model in models_reg.items():
    model.fit(X_train, yr_train)
    yp = model.predict(X_test)
    mse = mean_squared_error(yr_test, yp)
    r2 = r2_score(yr_test, yp)
    print(f"\n📊 {name} Regression:")
    print(f"  • MSE: {mse:.4f}")
    print(f"  • R²: {r2:.4f}")
    plt.figure(figsize=(5,4))
    sns.scatterplot(x=yr_test, y=yp)
    plt.xlabel("Actual Chance")
    plt.ylabel("Predicted")
    plt.title(f"{name} Predictions")
    plt.show()
```
